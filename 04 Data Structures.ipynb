````python
# generative expression
def generate_squares(n):
    """Generates squares of numbers from 0 up to n-1."""
    for i in range(n):
        yield i * i  # Pauses the function and yields the current square

# Using the generator
squares_gen = generate_squares(5) # Returns a generator object immediately
print(squares_gen) # Output: <generator object generate_squares at ...>

# Iterate through the generator to get values
print("Generated values:")
for square in squares_gen:
    print(square)
# or using next(square_gen)
````

## Stride
```python
sequence[start_index:end_index:stride]
```
- start_index: The starting position (inclusive).
- end_index: The stopping position (exclusive).
- stride: The number of elements to skip (interval).

s = "abcdefgh"
- Every second item	s[::2]	'aceg'
- Every third item	s[1:7:3]	'be'
- Default stride	s[0:4:1]	'abcd'
- Reverse the sequence	s[::-1]	'hgfedcba'
- Every second item, reversed	s[::-2]	'hfd'

````python
# Stride
names[::-1]
````

````python
list( range(10) )
````

````python
list( range(10) )[::3]
````

# Comprehensions
Common pattern, looping, mapping (optional), filtering, and accumulating

````python
# pattern for list comprehension
names2 = []
for name in names:
    if len(name) == 4:
        names2.append( name.title() ) # title is mapping
names2
````

````python
names2 = [name.title() for name in names if len(name) == 4]
names2
````

````python
# Dict Comprehensions
types = {'name': str, 'age': int, 'address': str}
````

````python
new_names = {}
for t in types:
    new_names[t] = t.title()
new_names
````

````python
new_names = { t: t.title() for t in types }
new_names
````

````python
names = ['john', 'paul', 'george', 'ringo']
````

````python
# Set Comprehensions
unique_names = { name for name in names if len(name) == 4 }
unique_names
````

````python
# Generator Expression
lazy_names = { name for name in names if len(name) == 4 }
lazy_names
````
