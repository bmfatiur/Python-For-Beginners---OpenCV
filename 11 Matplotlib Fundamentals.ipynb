# Matplotlib
## Figure & Axis Creation

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.arange(0, 10, 0.2)
y = np.sin(x)
fig = plt.figure()
ax = fig.add_subplot(111)
ax.plot(x, y)
plt.savefig("Files/pyplo1.png", dpi=300)
```

```python
from pylab import *
x = arange(0, 10, 0.2)
y = sin(x)
plot(x, y)
savefig("pylab1.png", dpi=300)
```

```python
fig = plt.figure()
fig.set_size_inches((8.5, 4))
```

```python
# single axes - position values are 0-1
left, bottom, width, height = 0.1, 0.2, 0.7, 0.5
ax = fig.add_axes( (left, bottom, width, height) )
ax.plot(x, y)

ax2 = fig.add_axes( (0.9, 0.2, 0.5, 0.2) )
ax2.plot(x, y)
fig
```

```python
# 3 axes - 1 row 3 cols (grid)
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
for ax in axes:
    ax.plot(x, y)
    x+=1
```

```python
# 1 axes - 1 row 2 cols 2nd position
ax = plt.subplot(122)
ax.plot(x, y)
```

```python
x = np.arange(0, 10, 0.2)
y = np.cos(x)
fig = plt.figure()
ax = fig.add_subplot(111)
ax.plot(x, y, color="r",
       linewidth = 3, linestyle = "--")
fig.savefig(Files/pyplot2.png, dpi=300)
```

# Plot Types
### Matplotlib supports a variety of plots out of the box

```python
# Line plot
fig, ax = plt.subplots()
ax.plot(x, y)
```

```python
# Bar Plot
fig, ax = plt.subplots(figsize=(10,8))
ax.bar(x, y)
```

```python
# Bar Plot
# width may need to be tweaked
fig, ax = plt.subplots(figsize=(10,8))
ax.bar(x, y, width=0.02)
```

```python
# Scatter Plot - using .scatter can be slower than plot. Use .scatter when you want to
# tweek attribute
fig, ax = plt.subplots(figsize=(10,8))
ax.scatter(x, y, marker='x', alpha=0.5)
```

```python
# Scatter Plot - test
fig, ax = plt.subplots(figsize=(10,8))

# with connecting line
ax.plot(x, y, marker='o', alpha=0.5, color='black', markeredgecolor='red', markerfacecolor='blue', markeredgewidth=3)

# without connecting line
# ax.scatter(x, y, alpha=0.5, color='black', edgecolors='red', linewidths=3)
```

```python
# Scatter Plot - Using .scatter can be slower than plot. Use .scatter when you want to
# tweak attribute
fig, ax = plt.subplots(figsize=(10, 8))
ax.scatter(x, y, marker='o', alpha=0.5)
```

```python
# Scatter Plot - Using .scatter can be slower than plot. Use .scatter when you want to
# tweak attribute
fig, ax = plt.subplots(figsize=(10, 8))
ax.scatter(x, y, marker='o', c=x, cmap='viridis', alpha=0.5)
```

```python
# Box Plot
fig, ax = plt.subplots(figsize=(10, 8))
_ = ax.boxplot(x, #vert=False
              )
```

```python
# boxplot
fig, ax = plt.subplots(figsize=(10, 8))
_ = ax.boxplot([x, x+2, x-3], labels=['Norm', '+2', '-3'])
```

```python
# Violin Plot
fig, ax = plt.subplots(figsize=(10, 8))
_ = ax.violinplot([x, x+2, x-3])
```

```python
# Histogram
fig, ax = plt.subplots(figsize=(10, 8))
_ = ax.hist(y)
```

```python
# Histogram
fig, ax = plt.subplots(figsize=(10, 8))
_ = ax.hist(y, bins=100)
```

```python
# Pie
fig, ax = plt.subplots(figsize=(10, 8))
_ = ax.pie([10, 5], labels=['10', '5'], autopct='%1.1f%%')
ax.legend()
```
