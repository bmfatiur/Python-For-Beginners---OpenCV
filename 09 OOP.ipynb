```python
class MyInt:
    '''Docstring for MyInt'''
    def __init__(self, val):
        self.value = val

    def __add__(self, other):
        return MyInt(self.value + other)

    def __repr__(self):
        return f'MyInt({self.value})'

    def __str__(self):
        return f'MyInt({self.value})'

    def square(self):
        "Return the square of the value"
        return MyInt(self.value**2)
```
