# Comprehensions
Common pattern, looping, mapping(optional), filtering(optional), and accumulating

```python
# pattern for list comprehension
namesComprehension = []
for name in names:
    if len(name)==4: # filtering
        namesComprehension.append(name.title()) # title is mapping
namesComprehension
```

```python
namesCopy = [name.title() for name in names if len(name)==4]
name
```

```python
namesCopy
```

```python
# Dictonary Comprehensions
types = {'name': str, 'age': int, 'address': str}
```

```python
namesDict = {}
for t in types:
    namesDict[t] = t.title()
namesDict
```

```python
# Dictonary Comprehension
temp = { t:t.title() for t in types }
temp
```

```python
# Set Comprehension
namesUnique = {name for name in names if len(name) == 4}
namesUnique
```

```python
# Generator Expression
namesLazy = ( name for name in names if len(name) == 4 )
namesLazy
```

```python
list(namesLazy)
```
