# NumPy Dimensions

```python
nums = np.arange(100).reshape(20, 5)
nums
```

```python
nums.transpose()
```

```python
# Axis - two dimensional
nums
```

```python
nums.mean()
```

```python
nums.mean(axis=0)
```

```python
nums.mean(axis=1, keepdims=True)
```

```python
# three dimensional
b = np.arange(70).reshape(7, 5, 2)
b
```

```python
b.mean(axis=0, keepdims=True)
```

```python
b.mean(axis=1, keepdims=True)
```

```python
b.mean(axis=2, keepdims=True)
```

```python
# NumPy Indexing & Slicing
# Similar to python, but not limited to one dimension
nums
```

```python
# row 0 only
nums[0]
```

```python
# rows 0, 5, 10
nums[[0, 5, 10]]
```

```python
# can slice along multiple dimensions:
nums[0:10] # first 10 rows
```

```python
# all rows, 3 columns
nums[:, 0:3]
```
