````python
# The ``with`` statement will automatically close your files. (Also used in plotting & setting pandas parameters
with open('Files/persons.csv', mode='w', encoding='utf8') as fout:
    fout.write('julia,31\n')
    fout.write('jenny,31\n')
````

### File opening modes
"w" (Write Mode): Use this mode when you want to create a new file or completely overwrite the contents of an existing file.\
"a" (Append Mode): Use this mode when you want to add new data to the end of an existing file without deleting the old data.
