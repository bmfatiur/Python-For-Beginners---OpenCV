### range
Python trends to follow half-open interval ( [start,end) ) with range and slices:

length = end - start

easy to concat ranges w/o overlap (ie list(range(3)) + list(range(3,9)) )

range(0, -10, -1) for -1 iteration

```python
# Java/C-esque style of object in array access (BAD):
animals = ["cat", "dog", "bird"]
for index in range(len(animals)):
    print(index, animals[index])
```

```python
# If you need indices, use ``enumerate`` ( to replace ``range(len(a_list))` )
animals = ["cat", "dog", "bird"]
for index, value in enumerate(animals):
    print(index, value)
```

```python
# break statement
animals = ["cat", "dog", "bird"]
for index, value in enumerate(animals):
    if value == "dog":
        break
    print(index, value)
```

```python
# continue statement
animals = ["cat", "dog", "bird"]
for index, value in enumerate(animals):
    if value == "dog":
        continue
    print(index, value)
```

```python
# Can loop over lists, strings, iterators, dictionaries... sequence-like things
my_dict = {"name": "matt", "cash": 5.45}
for key in my_dict: # loop over keys in default
    print(key)
```

```python
for value in my_dict.values():
    print(value)
```

```python
for key, value in my_dict.items():
    print(key, value)
```
