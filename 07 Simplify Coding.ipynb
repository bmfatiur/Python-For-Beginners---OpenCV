```python
def median(values):
    '''
    Return the middle value (if odd)
    or, the average of the two middle values (if even)
    >>> median([1, 4, 5])
        4
    >>> median([0, 2, 6, 100])
    4
    '''
    values = sorted( values )
    size = len( values )
    if size % 2 == 0:
        left = values[int(size/2 - 1)]
        right = values[int(size/2)]
        return (left+right)/2
    else:
        return values[ int(size/2) ]
```

```python
median
```

```python
median( range(100) )
```

```python
median( [100, 1, 200] )
```

```python
# Tuple - Return multiple items from a function
def roots(val):
    return (val**0.5, -(val**0.5))
```

```python
roots(4)
```

```python
# Lambda - One-line anonymous function
def adder(x, y):
    '''
    This adds two values
    >>> add(2, 4)
    6
    '''
    return x + y
```

```python
adder2 = lambda x, y: x + y
adder(42, 10) == adder2(42, 10)
```

```python
def roots(val):
    return (val**0.5, -(val**0.5))
```

```python
roots2 = lambda val: (val**0.5, -(val**0.5))
roots(64) == roots2(64)
```

```python
# Lambdas in sorting
names = ['john', 'paul', 'george', 'ringo', "David", "Mia", "Anthony", "Joe"]
```

```python
# alphabetically
sorted(names)
```

```python
# based on lengths
sorted(names, key=lambda name: len(name))
```
