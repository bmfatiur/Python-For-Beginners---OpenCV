```python
# Protected, Private, Getter, Setter
from datetime import datetime
class User:
    def __init__(self, username, email, password):
        self.username=username
        # Protected variable = _email
        # Private variable = __email
        self._email=email
        self.password=password
    def say_hi_to_user(self, user):
        print(f"Sending message from {self.username}: Hi {user.username}. How are you?")

    @property
    def email(self):
        print(f"Email accessed at {datetime.now()}")
        return self._email

    @email.setter
    def email(self, email):
        if "@" in email:
            print(f"Email edited at {datetime.now()}")
            self._email=email
        else:
            print("Enter valid email")
    

user1=User("Alice", "alice@mail.com", 123)
user2=User("Bob", "bob@mail.com", 321)

user1.say_hi_to_user(user2)
print(user1.email)
user1.email="uewyujsk@dms.com"
print(user1.email)
```

```python
# Abstract
class EmailService:
    def _connect(self):
        print("Connecting to Email server")
    def _authenticate(self):
        print("Authenticating to Email server")
    def send(self):
        self._connect()
        self._authenticate()
        print("Sending Email")
        self._disconnect()
    def _disconnect(self):
        print("Disconnecting from Email server")

gmail = EmailService()
gmail.send()
```

```python
# Inheritance
class Vehicle:
    def __init__(self, model, brand, year):
        self.model = model
        self.brand = brand
        self.year = year
    # def start(self):
    #     print(f"{self.model} is starting")
    # def stop(self):
    #     print(f"{self.model} is stopping")

class Car(Vehicle):
    def __init__(self, model, brand, year, number_of_doors, number_of_wheels):
        super().__init__(model, brand, year)
        self.number_of_doors = number_of_doors
        self.number_of_wheels = number_of_wheels
    def start(self):
        print("Car is starting")
    def stop(self):
        print("Car is stopping")

class Bike(Vehicle):
    def __init__(self, model, brand, year, number_of_wheels):
        super().__init__(model, brand, year)
        self.number_of_wheels = number_of_wheels
    def start(self):
        print("Bike is starting")
    def stop(self):
        print("Bike is stopping")

car = Car("Ford", "Focus", 2008, 5, 4)
bike = Bike("Honda", "Scoopy", 2018, 2)
print(car.__dict__)
print(bike.__dict__)

car.start()
car.stop()

bike.start()
bike.stop()
```

```python
# Polymorphism
class Plane(Vehicle):
    def __init__(self, model, brand, year, number_of_doors):
        super().__init__(model, brand, year)
        self.number_of_doors = number_of_doors
    def start(self):
        print("Plane is starting")
    def stop(self):
        print("Plane is stopping")
    
vehicles: list[Vehicle] = [
    Car("Ford", "Focus", 2008, 5, 4),
    Bike("Honda", "Scoopy", 2018, 2),
    Plane("Boeing", "747", 2015, 16)
]
for vehicle in vehicles:
    if isinstance(vehicle, Vehicle):
        print(f"Inspecting {vehicle.model} {vehicle.brand} ( {type(vehicle).__name__} )")
        vehicle.start()
        vehicle.stop()
```
