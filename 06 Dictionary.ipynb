```python
def median(values):
    """
    Return the middle value (if odds)
    or the average of the two middle values (if even)
    >>> median([1, 4, 5])
    4
    >>> median([6, 2, 100, 0])
    4.0
    """
    values = sorted(values)
    size = len(values)
    if size % 2 == 0:
        left = values[int(size/2) - 1]
        right = values[int(size/2)]
        return (left+right)/2
    else:
        return values[int(size/2)]
```

```python
# Lambda - One-line anonymous function
def adder(x, y):
    """
    This adds two values
    >>> add(2, 4)
    6
    """
    return x+y
```

```python
adderLambda = lambda x, y: x+y
adder(7, 9) == adderLambda(7, 9)
```
